```python
emp={"eid":101, "ename":"sayali","salary":30000}
print(emp)
```

```python
emp.keys() # sorts keys 
```

```python
emp.values() #sort values
```

```python
emp.popitem() #delets bydefault last key:value pair
print(emp)
```

```python
emp.pop("eid") #delete specific key:value pair which we passed as an arg
print(emp)
```

```python
print(emp)
```

```python
#how to add new key:value in existing dic
emp['eid']=1001
print(emp)
```

```python
#updating the existing key:value dic
emp.update({"ename":"tanu", "salary": 50000})
print(emp)
```

```python
emp["ename"]="xyz"
print(emp)
```

```python
marks={"python":90, "python": 80, "sql":50} #here same keys are not allowed and hence print 80 bcz dict is ordere
print(marks)
```

```python
Marks={"python":90, "sql":90, "powerbi":90, "tableau":68, "excel":90}
print(Marks)
```

```python
total_marks=sum(Marks.values())#we cant just write sum(Marks) bcz 
#it is combination of key:values so we only can calucate values 
print(total_marks)
```

```python
per=total_marks/len(Marks.values())
print(per)
```

```python
#coverting dict to list
sub=list(Marks.keys())
print(sub)
```

```python
Mrks=list(Marks.values())
print(Mrks)
```

```python
#converting list to dict using zip function
sub_Mrks=zip(sub,Mrks)
print(dict(sub_Mrks))
```

```python
#if we dont mention dict ie type cast it to dict then it returns object of the zip
sub_Mrks=zip(sub,Mrks)
print(sub_Mrks)
```

```python
#in python we cant pass more than 1 param to a datatype it throws error
#thats why we use zip function to compress it by ziping it into one var ie sub_Mrks
sub_Mrks=dict(sub,Mrks)
print(sub_Mrks)
```
