```python
#concating
import pandas as pd
a=pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['a','b','c','d','e']
})
b=pd.DataFrame({
    'id':[6,7,8],
    'name':['f','g','h']
})
```

```python
a=pd.concat([a,b],ignore_index=True)
```
