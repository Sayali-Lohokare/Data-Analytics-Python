```python
num=[1,2,2,3,3,3,4,3,5,6]
num=list(set(num)) #converting set to list and it will
# remove duplicates as list not allow duplicates
print(num)
```

```python
num = {23,11,1,2,3,4,5,6,7,8,9,9}
print(num)
```

```python
num.add(10)
print(num)
```

```python
num.add(100) #as set is unordered it will add number at any index irrespective of number count
print(num) 
```

```python
num.pop()#removes 1st element
print(num)
```

```python
num.remove(3) #here if we remove an element and then again try to remove it then it throws error
#so best option is to use "discard" with set
print(num)
```

```python
num.discard(2)
print(num)
```

```python
num.discard(2) # here even after discrading 2 we run the cell and still wont get error and can safely chcek the output
print(num)
```

```python
set1={1,2,3,4,5,6}
set2={5,6,7,8,9,10}
print(set1.intersection(set2)) #takes the same values from 2 sets
```

```python
set1.difference(set2) #pulls diff values from set
```

```python
set1.symmetric_difference(set2) #removes same val and gives whole set
```
