```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#create simple original img
image=np.zeros((28,28))
image[8:20,8:20]=1
```

```python
#add random noise

noise=np.random.normal(0,0.5,(28,28))
noise_image=image+noise
#noise_image
```

```python
#simple denoising
denoised_image=noise_image.copy()
for i in range(5):
    denoised_image+np.roll(denoised_image,1,axis=0)+np.roll(denoised_image,-1,axis=0)/3
```

```python
plt.figure(figsize=(10,3))
```

```python
plt.subplot(1,3,1)
plt.imshow(image,cmap='gray')
plt.axis('off')
```

```python
#after adding noise
plt.subplot(1,3,2)
plt.imshow(noise_image,cmap='gray')
plt.axis('off')
```

```python
plt.subplot(1,3,3)
plt.imshow(denoised_image,cmap='gray')
plt.axis('off')
```
