```python
import pandas as pd 
from sklearn.feature_extraction.text import CountVectorizer #convert text into nums
from sklearn.naive_bayes import MultinomialNB 
from sklearn.metrics import accuracy_score,confusion_matrix
 
# Sample dataset 
data = { 
    'text': [ 
        "Win money now!",  
        "Hi, how are you?", 
        "Congratulations, you've won a lottery!", 
        "Let's schedule a meeting tomorrow", 
        "Earn cash fast online", 
        "Reminder: your appointment is confirmed" 
    ], 
    'label': [1, 0, 1, 0, 1, 0]  # 1 = Spam, 0 = Not Spam 
} 
 
df = pd.DataFrame(data)
```

```python
df
```

```python
#convert text to nums
vector=CountVectorizer()
x=vector.fit_transform(df['text'])
x

#here text data is converted into int 64 bytes
```

```python
y=df['label']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.5,random_state=42)
```

```python
model=MultinomialNB()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
#kaggel dataset
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer #convert text into nums
from sklearn.naive_bayes import MultinomialNB 
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
df=pd.read_csv('spam.csv',encoding='latin-1')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum() #chceking null values 
```

```python
#converting spam n ham into numericals (0,1) using map() function
df['label']=df['Category'].map({'ham':0,'spam':1})
df.head()
```

```python
#feature selection
x=df['Message'] #kept in 1 dim bcz data is already in 2 dim ie in tabular form so kept it one dim
y=df['label']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python

vectorizer=CountVectorizer()#convert text into num
x_train_vector=vectorizer.fit_transform(x_train) #training using fit function
x_test_vector=vectorizer.transform(x_test) #testing
```

```python
model=MultinomialNB() #for oredicting num values
model.fit(x_train_vector,y_train)
```

```python
y_pred=model.predict(x_test_vector)
y_pred
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
confusion_matrix(y_test,y_pred)
```

```python
messages={
    'Congrats! You wona free lottery ticket',
    'Hey, are we meeting today?',
    'URGENT! You have won 1000 dollers',
    'plz call me when you reach home'
}
msg_vector=vectorizer.transform(messages)
```

```python
pred=model.predict(msg_vector)
pred
```

```python
for msg,pred in zip(messages,pred):
    if pred==1:
        result='SPAM'
    else:
        result='HAM'
    print(msg,":",result)
```

```python
#Bernolli NB-handles binary features(0,1)
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import BernoulliNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
messages=[
    'win free gifts',
    'free lottery win',
    'claim free prize',
    'metting 10pm tomrrow',
    'plz attend the meeting',
    'a free Europe trip voucher',
    'Congrats on your convocation ceremony'
    
]
y=[1,1,1,0,0,1,0]
```

```python
vectorizer=CountVectorizer(binary=True)
x=vectorizer.fit_transform(messages)
```

```python
#slpit featured into list
print('Features')
print(vectorizer.get_feature_names_out())
```

```python
#covert features into binary matrix
print("Binary matrix")
print(x.toarray()) #converting to array using toarray() function
#each row has a binary num for each feature and that followas by 7 col 
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=BernoulliNB()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
accuracy_score(y_pred,y_test)
```

```python
#Bayes thworem-after seeing some evidance how does the probability chnage tells BT
#formula- P(A|B)=P(B|A)*P(A)/P(B)
#where
#P(A)--prior probab
#P(B)--probab of evidance
#P(B|A)--probab of B when A is true
#P(A|B)--posterior probab -- what we want to find?
```

```python
30% of emails are spam
80% of spam emails contains the word 'free'
10% of non-spam email contain word 'free'
question---if an email contain word 'free' what is the probab that its is spam?
```

```python
p_spam=0.30
p_free_givenunder_spam=0.80
p_free_givenunder_notspam=10
p_notspam=0.70
#Q-calculate probability of seeling 'free'

#Formula--p('free')=p('free'/spam)p(spam)
p_free=(p_free_givenunder_spam*p_spam+p_free_givenunder_notspam*p_notspam)
print(p_free)
```

```python
p_spam_given_free=(p_free_givenunder_spam*p_spam)/p_free
print(p_spam_given_free)
```
