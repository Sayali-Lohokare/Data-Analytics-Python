```python
import pandas as pd
import numpy as np
```

```python
id=np.arange(1,21)
department=np.random.choice(['HR','Admin','Analyst','IT'],20)
gender=np.random.choice(['female','male'],20)
exp=np.random.randint(2,10,20)
age=np.random.randint(20,40,20)
salary=np.random.randint(20000,70000,20)
rating=np.round(np.random.uniform(1,5,20),1)
```

```python
data={
    'id':id,
    'department':department,
    'gender':gender,
    'exp':exp,
    'age':age,
    'salary':salary,
    'rating':rating
    
}
df=pd.DataFrame(data)
df
```

```python
#mathematical function
df['Total'] = df['Maths'] + df['Science'] 
df['Avg'] = df[['Maths','Science']].mean(axis=1) 
df['MaxVal'] = df[['A','B','C']].max(axis=1) 
df['Square'] = df['Score']**2
```

```python
df['salary hike']=df['salary']*1.1
df.head(4)
```

```python
df['Performance']=np.where(df['rating']>4,'Excellent','Average')
df.head(3)
```

```python
df['anual salary']=df['salary hike']*12
df.head()
```
