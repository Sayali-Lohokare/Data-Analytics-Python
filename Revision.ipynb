```python
#count of digit in number
#We used `str()` here because:

1. **Numbers aren't directly iterable** - You cannot loop through the digits of an integer directly with a `for` loop. If you tried `for d in num:`, Python would throw a TypeError.

2. **String conversion makes it iterable** - Converting the number to a string with `str(num)` transforms `12345` into `'12345'`, which is a sequence of characters that can be iterated over.

3. **Each character represents a digit** - When we loop through the string `'12345'`, each iteration gives us one character ('1', '2', '3', '4', '5'), allowing us to count each digit individually.

4. **Simple counting method** - This approach lets us count digits by simply counting the characters in the string representation, without needing mathematical operations like division or modulo.
num=12345
count=0
for d in str(num):
    count+=1
print(count)
```

```python
#chek if string is palindrome
def string(s): #usr-defind func
    if s==s[::-1]: #chcek if original and reverse s is same
        print("palindrome")
    else:
        print("not palindrome")
```

```python
#find duplicate in a list
def find(arr):
    s=set()
    duplicates=set()
    
    for x in arr:
        if x in s:
            duplicates.add(x)
        else:
            s.add(x)

    return list(duplicates)


find([1,2,3,5,3,2,1])
```
