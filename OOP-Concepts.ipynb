```python
class Student:
    def stud_info(self,rollno,name,cname):#when creating user-defined function in a class using self keyword is mandatory
        print(rollno,name,cname)

s=Student() #creating obj 
s.stud_info(101,"sayali","DS") #passing functions para to call function using obj
```

```python
class Student:
    def info(self,name,age,email): #method1
        print(name,age,email)

    def course(self,cname,fees,duration): #method2
        print(cname,fees,duration)

    def result(self,marks):  #method3
        print(marks)

s=Student()
s.info("sayali",24,"xyz@gmail.com")
s.course("DS",30000,"5 months")
s.result([67,45,88,91]) #passing list of marks

    
```
