```python
class Company:
    def __init__(self,cname):
        print(cname)

class Employee(Company):
    def __init__(self,cname,eid,ename,salary):
        super().__init__(cname)
        self.eid=eid #instance
        self.ename=ename
        self.salary=salary

    def emp_emp(self):
        print(self.eid,self.ename,self.salary)

e=Employee("TATA",101,"xyz",30000)
        
```

```python
e.emp_emp()
```

```python
class Login:
    def __init__(self,username,password):
        self.username=username
        self.password=password

    def login(self):
        print(self.username,self.password)

class Category(Login):
    def __init__(self,username,password,cname):
        super().__init__(username,password)
        self.cname=cname

    def category(self):
        print(self.cname)

class Product(Category):
    def __init__(self,username,password,cname,pname,price,qty):
        super().__init__(username,password,cname)
        self.pname=pname
        self.price=price
        self.qty=qty

    def product(self):
        print(self.pname,self.price,self.qty)

p=Product("abc","abc","gadgets","laptop",30000,1)
        
        
    
```

```python
p.login()
```

```python
p.category()
```

```python
p.product()
```

```python
#singlelevel inheritance
#here if we dont use super it will consider only child class pnae and hence shows no traces of amazon in outut
class Ecom:
    def __init__(self,ename):
        print(ename)

class Product(Ecom):
    def __init__(self,ename,pname,qty,price):
        self.pname=pname
        self.qty=qty
        self.price=price

    def show_product(self):
        print(self.pname,self.qty,self.price)

p=Product("amazon","laptop",1,5000)
```

```python
p.show_product()
```

```python
#singlelevel inheritance
#here we useed super() to call parent class product name and got amazon
#we create instance so that in differet method we can use those instance without creating new var always in diff methods
class Ecom:
    def __init__(self,ename):
        print(ename)

class Product(Ecom):
    def __init__(self,ename,pname,qty,price):
        super().__init__(ename)
        self.pname=pname
        self.qty=qty
        self.price=price

    def show_product(self):
        print(self.pname,self.qty,self.price)

p=Product("amazon","laptop",1,5000)
```

```python
p.show_product()
```

```python
class Course:
    def __init__(self,cname):
        print(cname)

class Subject(Course):
    def __init__(self,cname,sname,sid,duration,marks):
        super().__init__(cname)
        self.sname=sname
        self.sid=sid
        self.duration=duration
        self.marks=marks

    def show_details(self):
        print(self.sname,self.sid,self.duration,self.marks)

s=Subject("DS","Python",101,"30hrs",85)

        
    
```

```python
s.show_details()
```

```python
#when passing any default value in parameter it must be passed at the end of the parameter
class Course:
    def __init__(self,cname,fees=50000):
        print(cname,fees)

class Subjects(Course):
    def __init__(self,cname,sname,duration,fees=50000):
        super().__init__(cname,fees)
        self.sname=sname
        self.d=duration

    def show_subjects(self):
        print(self.sname,self.d)

s=Subjects("DS",["python","ml","maths","sql"],["1 month","1 month","15 days","15 days"])
s.show_subjects()   
```
