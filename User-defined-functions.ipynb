```python
def result(*marks,course="DS",rollno=1):
    print("course name:",course)
    print("Roll no:",rollno)
    print(marks)
    print("Total Marks:",sum(marks))
    per=sum(marks)/len(marks)
    print("Percentage",per)
    if per>=35:
        print("Pass")
    else:
        print("Fail")
```

```python
result(45,67,87,65,45,66)
```

```python
#Recursive function-funct calls itself
def summ(n):#halt1
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return n+summ(n-1)#halt2
summ(5)
```

```python
return--it is used to return avalue within the functionitself--recursion is 
happening within halt1&2 and return helps to take the input make a new
output and again move that output as new i/p and so on until condition fails

print()--it only prints values written inside it, it cant go into loop unlike return
```

```python
#factorial 
def fact(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return n*fact(n-1)
summ(5)
```

```python
def fact(n):
    if n==1 or n==0:
        return n
    else:
        return n*fact(n-1)
fact(5)
```

```python
#0,1,1,2,3,5
def fibo(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fibo(n-1)+fibo(n-2)
        #index 6-1+6
        #index 5+4
        #index 5+3
fibo(6)
```
