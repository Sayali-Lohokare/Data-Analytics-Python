```python
class Car:
    def start(self):
        print("car start")

class Bike:
    def start(self):
        print("bike start")

for v in (Car(),Bike()):
    v.start()
```

```python
#Duck type---single function use to call different class instea
class Cat:
    def sound(self):
        print("meow")

class Dog:
    def sound(self):
        print("bark")

def make_sound(animal):
        animal.sound()

make_sound(Cat())
make_sound(Dog())
```

```python
#genator---return items one at a time using yeild instead of return keyword, good for memory efficiency and for large data
def genator():
    for i in range(1,10):
        yield i

for i1 in genator():
     print(i1,end=" ")
```

```python
def read_file(filename):
    with open(filename,"r") as file:
        for line in file:
            yield line
for line in read_file("python.txt"): #this we copied path
    print(line)
```
