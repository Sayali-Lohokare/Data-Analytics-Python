```python
#multilevel inheritance-2/more classes allowed
class Data_analyst:
    def __init__(self,fees1,duration1):
        self.fees1=fees1
        self.duration1=duration1

    def show_DA(self):
        print(self.fees1,self.duration1)

class Data_Science:
    def __init__(self,fees2,duration2):
        self.fees2=fees2
        self.duration2=duration2

    def show_DS(self):
        print(self.fees2,self.duration2)

class Full_Stack:
    def __init__(self,fees3,duration3):
        self.fees3=fees3
        self.duration3=duration3

    def show_FS(self):
        print(self.fees3,self.duration3)

class AWS:
    def __init__(self,fees4,duration4):
        self.fees4=fees4
        self.duration4=duration4

    def show_AWS(self):
        print(self,fees4,duration4)

class Courses(Data_analyst,Data_Science,Full_Stack,AWS):
    def __init__(self,fees1,duration1,fees2,duration2,fees3,duration3,fees4,duration4):
        print("This is the course structure")
        Data_analyst.__init__(self,fees1,duration1)
        Data_Science.__init__(self,fees2,duration2)
        Full_Stack.__init__(self,fees3,duration3)
        AWS.__init__(self,fees4,duration4)

    def show(self):
        Data_analyst.show_DA(self)
        Data_Science.show_DS(self)
        Full_Stack.show_FS(self)
        AWS.show_AWS(self)

c=Courses(50000,"4months",60000,"6months",55000,"6months",50000,"6months")
```

```python
c.show()
```

```python
#hybrid inheritance--combination of multiple inheritance types
#student result
class student:
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno

    def display_stud(self):
        print("std roll no:",self.rollno,"\nstd name:",self.name)

class Academic(student):
    def __init__(self,name,rollno,marks):
        student.__init__(self,name,rollno)
        self.marks=marks

    def display_academic(self):
        print("marks:",self.marks)

class sport(student):
    def __init__(self,name,rollno,s_score):
        student.__init__(self,name,rollno)
        self.s_score=s_score

    def display_sport(self):
        print("sport score:",self.s_score)

class result(Academic,sport):
    def __init__(self,name,rollno,marks,s_score):
        Academic.__init__(self,name,rollno,marks)
        sport.__init__(self,name,rollno,s_score)

    def cal_result(self):
        return sum(self.marks)+self.s_score

    def display_result(self):
        self.display_stud()
        self.display_academic()
        self.display_sport()
        print("total marks:",self.cal_result())

r=result("sayali",101,[67,88,78,76,98,78],20)
r.display_result()
    
```
