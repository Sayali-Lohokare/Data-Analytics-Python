```python
it is online function without name
syntax: lambda arg: expression
```

```python
#square of a number
square=lambda a:a**2
print(square(10))
```

```python
print(square(2)) #allows using block of code again
```

```python
#addition
add=lambda a,b:a+b
print(add(6,7))
```

```python
print(add(56,66))
```

```python
is_even=lambda a: "even" if a%2==0 else "odd"

print(is_even(11))
```

```python
print(is_even(10))
```

```python
leap_year=lambda a: "leap year" if a%4==0 else "not leap year"
print(leap_year(11))
```

```python
#positive number or neg number
num=lambda a: "positive number" if a>0 else "negative number"
n=int(input("enter number"))
print(num(n))
```

```python
map fuction - used for calculation
filter function - use to filter data
```

```python
lst=[1,2,3,4,5,6,7,8]
square=map(lambda a:a**2,lst)
print(list(square))
```

```python
#changes letters to upper case
color=["yellow","blue","green"]
upper=map(lambda a: a.upper(),color)
print(list(upper))
```

```python
quanty=[10,20]
price=[100,300]
totalprice=map(lambda a,b:a*b,quanty,price) #using lambda map
print(list(totalprice))
```

```python
salary=[50000,40000,30000]
anual_salary=map(lambda a: a*12,salary)
print(list(anual_salary))
```

```python

salary_hike=map(lambda a: a*1.1,salary)
print(list(salary_hike))
```
