```python
#print even/odd num from list
lst=[1,2,3,4,5,6,5,7,6,10,8,9]
even=[]
odd=[]
for i in lst:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print(even)
print(odd)
```

```python
#while loop
pass: to prevent error(means pass condition/loop without execution)
break: to stop execution
continue: to skip the specific condition or value
```

```python
for i in range(1,10):
    if i==5:
        break
    print(i)
```

```python
for i in range(1,10):
    if i==5:
        continue
    print(i)
```

```python
for i in range(1,10):
    if i==5:
        pass
    print(i)
```
