**When to use a tuple instead of a list**

- Data that shouldn't change — coordinates, RGB colors, database records.

- Safety — immutability protects data from accidental edits.

- As dictionary keys — lists can't be keys, but tuples can (they're hashable).

- Slightly faster and lighter than lists.

**Use of packing n unpacking**

1. Swapping variables. Two values can be exchanged in a single statement without a temporary variable. The right side is packed into a tuple and then unpacked into the variables on the left.
Example: x, y = y, x

2. Returning multiple values from a function. A function can return only one object, but that object can be a tuple containing several values. The caller then unpacks it into separate variables.
Example: return min_val, max_val (packing), and low, high = function() (unpacking).

3. Assigning multiple variables in one line. Several variables can be given values together, which makes the code shorter and easier to read.
Example: name, age, city = "Sayali", 22, "Pune"

4. Handling sequences of unknown length. The * operator collects the remaining values into a list, so the first few items can be separated from the rest.
Example: first, *rest = (1, 2, 3, 4)

5. Iterating over pairs or nested data. Each inner list or tuple can be unpacked directly in a for loop, which gives meaningful variable names instead of index numbers.
Example: for name, marks in students:

6. Working with enumerate() and dictionaries. Functions such as enumerate() and dict.items() produce pairs of values, and unpacking separates each pair into two variables.
Example: for key, value in d.items():

**set operations  - union, intersection, difference**

- union -- > symbol ( | ) --> a.union(b) --> all items from both

- intersection --> symbol( & ) --> a.intersection(b) --> items in both

- difference --> symbol( - ) --> a.difference(b) --> in a but not b

- symmetric diff --> symbol( ^ ) --> a.symmetric_diffrenece(b) --> not shared by both

```python
j={1,2,3,4}
g={3,4,5,6}
```

```python
# union

print(j|g)
print(j.union(g))
print(g.union(j))
```

```python
# intersection

print(j&g)
print(j.intersection(g))
```

```python
# difference

print(j - g) # in j but not in g
print(g-j) # in g but not in j
print(j.difference(g))
print(g.difference(j))
```

```python
# symmetric diff

print(j^g)
print(g^j)
print(j.symmetric_difference(g))
print(g.symmetric_difference(j))
```

**namedtuple**

- plain tuple works by position, we have to remember what each index means which gets confusing
- namedtuple() allows to access fields by name, eg std.gpa instead of std[index]

```python
#using plain tuple

std=('amit',21,'bca',8.4)

print(std[0])
print(std[3])
```

- namedtuple comes from collections module.
- we can define 'shape' once then create records from it
- it will still remain a tuple - indexable,immutable but now readable

```python
from collections import namedtuple

#defining shape : name + its fields
stud=namedtuple('student',['name','age','course','gpa'])

#create records
s=stud('sayali',24,'MSc',9.87) # s is obj here

print(s.name)
print(s.age)
print(s.gpa)
print(s[0]) #still can work like tuple
```

**Methods of namedtuple()**

- _make() - build a record from a list/tuple of vaues, no need to write stud(row[0],row[3])
- _asdict() - converts tuple to dict, handy for converting to JSON for an API response
- _replace(field=value) - make new record with asked field replaced, Tuples are immutable, so you can't edit in place. _replace() return a fresh record with ur change. It never change the original tuple, it returns a new one
- _fields - see all field names(a tuple), used for looping over fields or printing header row for a table

**When to use namedtuple()**

- fixed records - a std, a point, an RGB color, a database row
- when index num get confusing - names make code readable
- lightweight alternative to a class - when u just need to hold data n not behaviour
- data from rows - _make() turn csv/db rows into clean record

**Rule of thumb: If you're using a plain tuple and keep forgetting what each position
means — switch to a named tuple.**

```python
# _make()

stud=namedtuple('student',['name','age','course','gpa'])

row=['neha',24,'mba',9.88]
s=stud._make(row) #turn list into a student
print(s.name,s.gpa)
```

```python
# _asdict()

d=s._asdict()
print(d)
```

```python
# replace(field=value)

new=s._replace(gpa=9.2) # replace the required field and make a new tuple
print(new)
print(s.name,s.age,s.course,s.gpa) # the original tuple still has same gpa
```

```python
# _fields()

print(s._fields)
```
