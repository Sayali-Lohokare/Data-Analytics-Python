```python
# Simple brute force approach - no fancy libraries needed!

# Step 1: Create empty lists for each column
id_list = []
dep_list = []
gender_list = []
experience_list = []
salary_list = []
age_list = []
rating_list = []

# Step 2: Define our options
departments = ['HR', 'IT', 'Finance', 'Marketing', 'Operations']
genders = ['Male', 'Female']

# Step 3: Use a simple loop to create 100 employees
import random

for i in range(100):  # Start small with 100 employees
    # Add ID (just counting 1, 2, 3...)
    id_list.append(i + 1)
    
    # Pick random department
    dep_list.append(random.choice(departments))
    
    # Pick random gender
    gender_list.append(random.choice(genders))
    
    # Pick random experience (0 to 25 years)
    experience_list.append(random.randint(0, 25))
    
    # Pick random salary (30000 to 100000)
    salary_list.append(random.randint(30000, 100000))
    
    # Pick random age (22 to 60)
    age_list.append(random.randint(22, 60))
    
    # Pick random rating (1.0 to 5.0)
    rating_list.append(round(random.uniform(1.0, 5.0), 1))

# Step 4: Print first 5 employees to check
print("First 5 employees:")
for i in range(5):
    print(f"ID: {id_list[i]}, Dept: {dep_list[i]}, Gender: {gender_list[i]}")
    print(f"Experience: {experience_list[i]}, Salary: {salary_list[i]}")
    print(f"Age: {age_list[i]}, Rating: {rating_list[i]}")
    print("-" * 30)
```

```python
#Filtering/Boolean indexing

arr=np.array([34,87,67,27,19,45,90,55,11])
arr[arr>50] #prints array with values >50
```

```python
arr[(arr%2==0) & (arr>20)] #it chceks of values in arr is divisible by 2 and return those even values >20
# also AND check if both RHS and LHS conditions are true and only then gives result
```

```python
arr[(arr%2==0) | (arr>20)] #here OR operation check if any of the either condition is true if yes then gives that output
```
