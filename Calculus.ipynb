derivative

```python
x=sp.symbols('x')
y=3*x**2+2*x+5

dy=sp.diff(y,x)
print(dy)
```

```python
dy.subs(x,2)
```

Partial Derivative

```python
x,y=sp.symbols('x y')
z=x**2+y**2
dx=sp.diff(z,x)
dy=sp.diff(z,y)
```

```python
print(dx,dy)
```

Gradient

```python
z=x**2+y**2
gradient=[sp.diff(z,x),sp.diff(z,y)]
print(gradient) #all partial serivatives together
```

GRADIENT DESCENT

```python
w=0 #parameter
learning_rate=0.1
for i in range(20):

    #cost function
    cost=(w-5)**2

    #derivative direction of change
    gradient=2*(w-5)

    #update w
    w=w-learning_rate*gradient

    print(i,'w=',w,'Cost=',cost)
```

Integral Calculus

```python
import sympy as sp
x=sp.symbols('x')
f=x**2

integral=sp.integrate(f,x)
print(integral)
```
